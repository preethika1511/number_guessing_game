```python
import random

print("Welcome to the Guess the number game:")
rand=random.randint(1,100)
while True:
        
        try:
            number=int(input("Guess the number between 1 to 100:"))
            if (number>rand):
                print("Too high!")  
            elif (number<rand):
                print("too low!")   
            else:
                print("Congrats!You guessed the correct number.")
                break
        except:
            print("Please enter a valid number!")
            continue
```
